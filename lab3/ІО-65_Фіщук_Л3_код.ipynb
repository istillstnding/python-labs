```python
input_string = input("Введіть послідовність символів: ")

result_chars = []

for char in input_string:
    if char == 'я':
        result_chars.append(' ')
    else:
        next_char = chr(ord(char) + 1)
        result_chars.append(next_char)

result_string = "".join(result_chars)

print(f"Оброблений рядок: {result_string}")
```

```python
text = input("Введіть послідовність символів: ")
byte_text = bytearray(text, encoding="utf-8")

words = byte_text.split(b' ')

longest_word = max(words, key=len)

longest_word_str = longest_word.decode("utf-8")
word_len = len(longest_word_str)

print(f"Найдовше слово: {longest_word_str}, його довжина: {word_len}")
```
