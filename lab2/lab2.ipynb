```python
import math

x = float(input("Введіть значення x: "))
y = float(input("Введіть значення y: "))
z = float(input("Введіть значення z: "))
a = float(input("Введіть значення a: "))
b = float(input("Введіть значення b: "))

numerator = x + 2.6 * y * math.sin(math.radians(z))
denominator = a - b ** 3

if numerator < 0:
    print("Значення чисельника не відповідає ОДЗ")
elif denominator == 0:
    print("Значення знаменника не відповідає ОДЗ")
else:
    print("Значення виразу: ", math.sqrt(numerator) / denominator)
```

```python
import math

s1 = float(input("Введіть площу кола: "))
s2 = float(input("Введіть площу квадрата: "))

d_circle = 2 * math.sqrt(s1 / math.pi)
a_square = math.sqrt(s2)
d_square = a_square * math.sqrt(2)

if math.isclose(d_circle, a_square):
    print("Коло вписане в квадрат")
elif d_circle <= a_square:
    print("Коло поміщається в квадрат")
else:
    print("Коло не поміщається в квадрат")

if math.isclose(d_circle, d_square):
    print("Квадрат вписаний в коло")
elif d_square <= d_circle:
    print("Квадрат поміщається в коло")
else:
    print("Квадрат не поміщається в коло")
```

```python
import math

n = int(input("Введіть число (>100): "))

if n < 100:
    print("Число має бути більше за 100")
else:
    range1 = range(10, n + 1)

    for i in range1:
        square = math.isqrt(i)
        if square * square == i:
            print(f"Число: {i}, Квадратний корінь: {square}")

    for i in range1:
        root = round(math.cbrt(i))
        if root ** 3 == i:
            print(f"Число: {i}, Кубічний корінь: {root}")
```
